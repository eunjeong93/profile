```python
from flask import Flask, jsonify, request
import networkx as nx
import pickle
import json
import numpy as np
from flask_cors import CORS
from pyvis.network import Network

with open("graph.pkl", "rb") as f:
    G = pickle.load(f)

# 🔹 Pyvis 네트워크 그래프 생성 (배경 검정, 글씨 흰색)
net = Network(notebook=True, height="700px", width="100%",
              bgcolor="#222222", font_color="white")

# 🔹 networkx 그래프 → pyvis 변환
net.from_nx(G)

# 🔹 노드 속성 추가 (툴팁에 노드 정보 표시)
for node, data in G.nodes(data=True):
    net.get_node(node)["title"] = str(data)  # 노드 설명 추가
    net.get_node(node)[
        "color"] = "lightblue" if data["type"] == "user_name" else "lightcoral"

# 🔹 HTML 파일로 저장 & 실행
net.show("graph.html")  # 브라우저에서 실행 가능
print("✅ 그래프가 `graph.html`로 저장되었습니다. 브라우저에서 열어보세요!")

```
